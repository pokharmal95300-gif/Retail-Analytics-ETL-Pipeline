# Retail Analytics ETL Pipeline — Simple Local Python Version

This notebook runs in ordinary Jupyter Notebook on your laptop. It does **not** require PySpark, Hadoop, Java, Azure, or `winutils.exe`.

Keep `sales_data.csv` and `footfall_data.csv` in the same Jupyter folder as these notebooks (your screenshot shows both files in the Jupyter root folder).

Run the notebooks in this order: **Bronze → Silver → Gold**.

## Bronze Layer
Copy the original sales and footfall CSV files into the local Bronze output folder. The original source files are not modified.


```python
from pathlib import Path
import csv
import shutil

# Find CSV files in the current Jupyter folder or its data subfolder
BASE_DIR = Path.cwd()

def find_csv(filename):
    candidates = [
        BASE_DIR / filename,
        BASE_DIR / "data" / filename,
        BASE_DIR.parent / filename
    ]
    for path in candidates:
        if path.exists():
            return path
    raise FileNotFoundError(
        f"{filename} not found. Put it beside this notebook or inside a 'data' folder. "
        f"Current folder: {BASE_DIR}"
    )

sales_file = find_csv("sales_data.csv")
footfall_file = find_csv("footfall_data.csv")

print("Sales file:", sales_file)
print("Footfall file:", footfall_file)
```

```python
# Check that both files have the required columns
required_sales = {"Receipt_ID", "Store_ID", "Timestamp", "Total_Amount"}
required_footfall = {"Store_ID", "Timestamp", "Footfall_Count"}

with open(sales_file, newline="", encoding="utf-8-sig") as f:
    sales_columns = set(csv.DictReader(f).fieldnames or [])

with open(footfall_file, newline="", encoding="utf-8-sig") as f:
    footfall_columns = set(csv.DictReader(f).fieldnames or [])

if not required_sales.issubset(sales_columns):
    raise ValueError(f"Sales CSV columns missing: {required_sales - sales_columns}")

if not required_footfall.issubset(footfall_columns):
    raise ValueError(f"Footfall CSV columns missing: {required_footfall - footfall_columns}")

print("Both CSV files have the correct columns.")
```

```python
# Save source files into Bronze layer
output_root = BASE_DIR / "retail_analytics_output"
bronze_dir = output_root / "bronze"
bronze_dir.mkdir(parents=True, exist_ok=True)

shutil.copy2(sales_file, bronze_dir / "sales_data.csv")
shutil.copy2(footfall_file, bronze_dir / "footfall_data.csv")

print("Bronze layer created successfully.")
print("Sales rows:", sum(1 for _ in open(sales_file, encoding="utf-8-sig")) - 1)
print("Footfall rows:", sum(1 for _ in open(footfall_file, encoding="utf-8-sig")) - 1)
print("Output folder:", bronze_dir)
```
